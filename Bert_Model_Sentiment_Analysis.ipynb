```python
from transformers import BertTokenizer, BertForSequenceClassification
from transformers import pipeline

# Load pre-trained model and tokenizer
model_name = "nlptown/bert-base-multilingual-uncased-sentiment" #model_name = "distilbert-base-uncased-finetuned-sst-2-english"  # Another pre-trained model for sentiment analysis

tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertForSequenceClassification.from_pretrained(model_name)

# Create a sentiment analysis pipeline
sentiment_analyzer = pipeline("sentiment-analysis", model=model, tokenizer=tokenizer)
```

```python
# Example reviews
reviews = [
    "I love this product! It has changed my life.",
    "The quality is terrible, I'm very disappointed.",
    "It's okay, not great but not bad either.",
    "Excellent service and fast shipping!",
    "Not worth the money. Very poor performance."
]
```

```python
# Analyze sentiment for each review
results = sentiment_analyzer(reviews)

# Print results
for review, result in zip(reviews, results):
    print(f"Review: {review}\nSentiment: {result['label']} (Score: {result['score']:.4f})\n")
```

# **Full code**

```python
from transformers import BertTokenizer, BertForSequenceClassification, pipeline
import pandas as pd

# Load pre-trained model and tokenizer
model_name = "nlptown/bert-base-multilingual-uncased-sentiment"
tokenizer = BertTokenizer.from_pretrained(model_name)
model = BertForSequenceClassification.from_pretrained(model_name)

# Create a sentiment analysis pipeline
sentiment_analyzer = pipeline("sentiment-analysis", model=model, tokenizer=tokenizer)

# Sample DataFrame setup for demonstration
data = {
    'cleaned_review_text': [
        "I love this product! It has changed my life.",
        "The quality is terrible, I'm very disappointed.",
        "It's okay, not great but not bad either.",
        "Excellent service and fast shipping!",
        "Not worth the money. Very poor performance."
    ]
}
df = pd.DataFrame(data)

# Function to analyze sentiment using the Hugging Face pipeline
def analyze_sentiment(review_text):
    result = sentiment_analyzer(review_text)[0]
    return f"{result['label']} (Score: {result['score']:.4f})"

# Apply sentiment analysis to the dataset
df['sentiment'] = df['cleaned_review_text'].apply(analyze_sentiment)

df.to_csv('sentiment_analysis_results.csv', index=False)
```
